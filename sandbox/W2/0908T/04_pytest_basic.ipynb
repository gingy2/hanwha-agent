### Pytest
내가 만든 프로그램이 예상대로 동작하는지 자동으로 검사해주는 도구

- pytest 에는 “이 테스트를 실행해라” 라고 등록하는 곳이 없다. ⇒ **이름 규칙이 곧 등록이다.**
    - 파일 : `test_*.py`
    - 함수 : 이름이 `test` 로 시작

```python
import os                   # 운영체제 관련 표준 라이브러리
import pathlib              # 파일 경로를 객체로 다루는 표준 라이브러리

here = pathlib.Path.cwd()   # 현재 작업 위치 확인

ROOT=here.parents[2] if here.name=='0907M' else here    # hanwha-agent

os.chdir(ROOT)  # 앞으로 모든 상대경로는 이 폴더가 기준이 된다.
```
